```python
import matplotlib.pyplot as plt
import networkx as nx

# Erstellen Sie ein gerichtetes Graphenobjekt
G = nx.DiGraph()

# Fügen Sie Knoten für jede Komponente Ihrer Anwendung hinzu
components = ['Mobile App (Flutter)', 'Backend', 'Database', 'Chatbot', 'Wikibot']
G.add_nodes_from(components)

# Fügen Sie Kanten zwischen den Komponenten hinzu, um ihre Beziehungen darzustellen
edges = [
    ('Mobile App (Flutter)', 'Backend'),
    ('Mobile App (Flutter)', 'Chatbot'),
    ('Backend', 'Database'),
    ('Backend', 'Chatbot'),
    ('Chatbot', 'Backend'),
 #   ('Website', 'Backend'),
  #  ('Website', 'Chatbot'),
    ('Wikibot', 'Backend'),
]
G.add_edges_from(edges)

# Zeichnen Sie das Diagramm
plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G)  # Positionierung der Knoten mit dem Fruchterman-Reingold-Algorithmus
nx.draw(G, pos, with_labels=True, node_size=3000, node_color='lightblue', font_size=10, font_weight='bold', edge_color='gray', width=2, alpha=0.6)

# Anzeigen des Diagramms
plt.title('Software Architecture')
plt.show()
```
