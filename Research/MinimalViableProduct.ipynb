```python
import matplotlib.pyplot as plt
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
import numpy as np
import requests
from PIL import Image

def create_organizational_chart_goals_and_paths():
    # Define the structure of the chart with shortened labels
    positions = {
        "10-Year Goal": (0, 5),
        "5-Year Pathway 1": (-2, 4), "5-Year Pathway 2": (2, 4),
        "3-Year Milestone 1": (-2, 3), "3-Year Milestone 2": (2, 3),
        "1-Year Plan 1": (-2, 2), "1-Year Plan 2": (2, 2),
        "Objective 1": (-2, 1), "Objective 2": (2, 1),
        "Key Result 1": (-2, 0), "Key Result 2": (2, 0)
    }

    # Create the plot
    fig, ax = plt.subplots()
    ax.set_xlim(-3, 3)
    ax.set_ylim(-1, 6)
    plt.axis('off')

    # Plot the positions
    for position, (x, y) in positions.items():
        ax.text(x, y, position, ha='center', va='center', bbox=dict(facecolor='white', alpha=0.5))

        # Add lines to connect the nodes
        if "10-Year Goal" not in position:
            # Determine parent position
            parent_x, parent_y = positions["10-Year Goal"] if "5-Year Pathway" in position else (x, y + 1)
            ax.plot([x, parent_x], [y, parent_y], 'gray')

    # Display the chart
    plt.show()

create_organizational_chart_goals_and_paths()
```
